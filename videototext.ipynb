```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled, NoTranscriptFound

def get_video_transcript(video_id, languages=('en', 'hi')):
    """
    Fetches the transcript for the given YouTube video ID in preferred languages.
    """
    for lang in languages:
        try:
            transcript = YouTubeTranscriptApi.get_transcript(video_id, languages=[lang])
            transcript_text = ' '.join([entry['text'] for entry in transcript])
            print(f"Transcript found in language: {lang}")
            return transcript_text
        except (TranscriptsDisabled, NoTranscriptFound):
            print(f"No transcript available in language: {lang}")
        except Exception as e:
            print(f"Error fetching transcript in {lang}: {e}")
    return None

def summarize_text(text, max_length=150):
    from transformers import pipeline

# Initialize the summarization pipeline on CPU
    summarizer = pipeline("summarization", model="facebook/bart-large-cnn", device=-1)
    max_chunk_size = 1000
    chunks = [text[i:i+max_chunk_size] for i in range(0, len(text), max_chunk_size)]
    summary = ""
    for chunk in chunks:
        try:
            summary_chunk = summarizer(chunk, max_length=max_length, min_length=30, do_sample=False)[0]['summary_text']
            summary += summary_chunk + " "
        except Exception as e:
            print(f"Error during summarization: {e}")
    return summary.strip()

if __name__ == "__main__":
    video_id = "jQo6n-i6wpo&list"  # Replace with your video ID

    transcript_text = get_video_transcript(video_id, languages=('en', 'hi'))
    if transcript_text:
        print("Transcript fetched successfully.")
        summary = summarize_text(transcript_text)
        print("Summary of the video:")
        print(summary)
    else:
        print("Could not retrieve a transcript in the preferred languages.")
```
